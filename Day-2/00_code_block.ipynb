# Upload & Extract `IIMT` folder to Google Drive

Run the single cell below. It will:
1. Mount your Google Drive
2. Open a file picker so you can select `IIMT-20260725T230040Z-1-001_1_.zip` from your computer
3. Extract it directly into `MyDrive/` so it lands at `MyDrive/IIMT/...`
4. Clean up the temporary zip from the Colab runtime

After this, `Day-2.ipynb` will find everything at `/content/drive/MyDrive/IIMT/` as expected.

```python
from google.colab import drive
from google.colab import files
import zipfile, os

# 1. Mount Google Drive (skips if already mounted)
drive.mount('/content/drive')

# 2. Prompt file picker -- select your IIMT zip here
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# 3. (Optional) uncomment to wipe any existing IIMT folder first for a clean copy
# import shutil
# shutil.rmtree('/content/drive/MyDrive/IIMT', ignore_errors=True)

# 4. Extract directly into MyDrive (zip already contains a top-level IIMT/ folder)
with zipfile.ZipFile(filename, 'r') as z:
    z.extractall('/content/drive/MyDrive/')

# 5. Clean up the zip from the Colab runtime (Drive keeps only the extracted folder)
os.remove(filename)

print("Done. Files are now at /content/drive/MyDrive/IIMT/")
```
